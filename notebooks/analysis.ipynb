# Results analysis
This notebook only analyses outputs. All training runs come from `src/train.py` with a config file.
Run `make eval` first so the tables in `reports/tables/` exist.

```
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from pathlib import Path
T = Path('../reports/tables')
yt = pd.read_csv(T/'results_by_year.csv'); dm = pd.read_csv(T/'dm_tests.csv'); reg = pd.read_csv(T/'regimes.csv')
```

## Pooled results and DM tests vs HAR

```
yt[yt.test_year=='pooled'].sort_values('qlike')
```

```
dm[dm.test_year=='pooled'].pivot(index='model', columns='loss', values='p_value').round(3)
```

## Per-year QLIKE relative to HAR

```
per = yt[yt.test_year!='pooled'].pivot(index='test_year', columns='model', values='qlike')
(per.div(per['HAR'], axis=0)).round(3).style.background_gradient(cmap='RdYlGn_r', vmin=0.8, vmax=1.2)
```

## Regimes (V5)

```
reg.pivot(index='model', columns='regime', values='qlike_vs_har').round(3)
```

## Failure cases
The worst forecasts by QLIKE for each deep model. Look for under-prediction at volatility jumps, e.g. earnings or March 2020.
